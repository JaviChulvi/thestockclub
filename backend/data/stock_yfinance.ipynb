```python
import yfinance as yf
import pandas as pd
import numpy as np
# import matplotlib.pyplot as plt
# import seaborn as sns
```

```python
def get_stock_price_history(ticker: str):
    stock = yf.Ticker(ticker)
    return stock.history(period="1y", interval="1d", auto_adjust=False)

def get_stock_info(ticker: str):
    stock = yf.Ticker(ticker)
    return stock.get_info()

def clean_stock_info(stock_info: dict):
    # Define the keys we want to keep based on the categories
    keys_to_keep = {
        # General info
        'country',
        'sectorKey',
        'industryKey',
        'longBusinessSummary',
        'fullTimeEmployees',
        'address1',
        'website',
        'currency',
        'exchange',
        # Valuation & Market Size
        'symbol',
        'marketCap',
        'currentPrice',
        'trailingPE',
        'forwardPE',
        'priceToBook',
        'beta',
        'totalCash',
        'totalDebt',
        'currentRatio',
        'debtToEquity',
        
        # Profitability & Cash Flow
        'totalRevenue',
        'netIncomeToCommon',
        'profitMargins',
        'returnOnEquity',
        'freeCashflow',
        'ebitda',
        
        # Growth & Performance
        'revenueGrowth',
        'earningsGrowth',
        'trailingEps',
        'fiftyTwoWeekHigh',
        'fiftyTwoWeekLow',
        
        # Financial Health & Risk
        'totalCash',
        'totalDebt',
        'debtToEquity',
        'currentRatio',
        'overallRisk',
        
        # Analyst Expectations
        'recommendationKey',
        'targetMeanPrice',
        'numberOfAnalystOpinions'
    }
    
    # Filter the stock_info to only include the keys we want and exclude None values
    return {k: v for k, v in stock_info.items() if k in keys_to_keep and v is not None}

```
